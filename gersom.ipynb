```python
  # Mengimpor pustaka NumPy untuk operasi matriks dan array numerik
  import numpy as np

  # Menampilkan pesan konfirmasi bahwa library berhasil diimpor
print("Library NumPy berhasil diimpor.")
```

```python
# Mendefinisikan matriks koefisien A (ukuran 3x3) dengan tipe data floating point (desimal)
A = np.array([
    [2.0,  1.0, -1.0],
    [-3.0, -1.0,  2.0],
    [-2.0,  1.0,  2.0]
  ], dtype=float)

# Mendefinisikan vektor konstanta b (ukuran 3x1) dari hasil persamaan SPLTV
b = np.array([8.0, -11.0, -3.0], dtype=float)

# Mencetak isi dari Matriks A dan Vektor b ke layar
print("Matriks A:\n", A)
print("Vektor b:\n", b)
```

```python
# Membuat fungsi Eliminasi Gauss untuk menyelesaikan SPLTV
def gauss_elimination_spltv(A, b):
    n = len(b)
    M = np.column_stack((A, b))   # Menggabungkan matriks A dan b (augmented matrix)

    # 1. Eliminasi Maju: Mengubah matriks menjadi segitiga atas menggunakan OBE & Pivoting
    for i in range(n):
        max_row = i + np.argmax(np.abs(M[i:, i]))
        if M[max_row, i] == 0:
            raise ValueError("Matriks tidak memiliki solusi unik.")
        if max_row != i:
            M[[i, max_row]] = M[[max_row, i]]

        for j in range(i + 1, n):
            factor = M[j, i] / M[i, i]
            M[j, i:] -= factor * M[i, i:]

    # 2. Substitusi Balik: Menghitung nilai variabel x, y, dan z dari bawah ke atas
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (M[i, -1] - np.sum(M[i, i+1:n] * x[i+1:n])) / M[i, i]
    return x

# Menjalankan fungsi dan mencetak hasil akhir (x, y, z)
solusi = gauss_elimination_spltv(A, b)
print("HASIL PENYELESAIAN:")
print(f"x = {solusi[0]:.4f}")
print(f"y = {solusi[1]:.4f}")
print(f"z = {solusi[2]:.4f}")
```
